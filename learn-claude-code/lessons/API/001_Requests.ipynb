```python
# Create an API client
from anthropic import Anthropic
client = Anthropic()
model = "claude-sonnet-5" 
```

```python
# Make a request to the API
import anthropic

# Initialize the client (automatically looks for ANTHROPIC_API_KEY env variable)
client = anthropic.Anthropic()

response = client.messages.create(
    model=model,                          # Required: Target model
    max_tokens=1024,                     # Required: Max output tokens
    messages=[                           # Required: Array of message objects
        {
            "role": "user",
            "content": "Hello, Claude!"
        }
    ]
)

# response.content can include a ThinkingBlock before the TextBlock
# (adaptive thinking is on by default on this model), so find the
# first text block instead of assuming content[0] is text.
print(next(block.text for block in response.content if block.type == "text"))
```

```python
def add_user_message(messages, text):
    user_message = {
        "role": "user",
        "content": text
    }
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {
        "role": "assistant",
        "content": text
    }
    messages.append(assistant_message)

def chat(messages):
    response = client.messages.create(
        model=model,
        max_tokens=1024,
        messages=messages
    )
    # response.content can include a ThinkingBlock before the TextBlock
    # (adaptive thinking is on by default on this model), so find the
    # first text block instead of assuming content[0] is text.
    return next(block.text for block in response.content if block.type == "text")   
```

```python
# Make a start listing of messages
messages = []
# Add in the initial user message
add_user_message(messages, "Define Quantum Computing in one sentence.")

# Pass the message to chat() to get a response from the model
chat_response = chat(messages)
# Take the answer and add it to the messages list
add_assistant_message(messages, chat_response)
# Add in the user followup message
add_user_message(messages, "What are some potential applications of Quantum Computing?")
# Call chat() again with the list of messages to get the final answer from the model 
final_response = chat(messages)
```
